```python
import sys
import os
import numpy as np
import pandas as pd

sys.path.append(os.path.abspath("../src"))

from naive_bayes import SpamDetectorNB
```

```python
# Open the dataset and count the number of spam/ham mails
mails = pd.read_csv("../data/spamham.csv")

count = mails['Category'].value_counts()
print(count)
```

```python
# Split data as training and testing sets
mask = np.random.rand(len(mails)) < 0.8
training_set = mails[mask]
testing_set = mails[~mask]

X_train = training_set["Message"]
y_train = training_set["Category"]

X_test = testing_set["Message"]
y_test = testing_set["Category"]
```

```python
# Train the model
model = SpamDetectorNB()

model.train(X_train, y_train)
```

```python
# Predict the category of the training set
y_pred = model.predict(X_test)
```

```python
# Compute the accuracy
accuracy = sum(
    y_pred[i] == y_test.iloc[i]
    for i in range(len(y_test))
) / len(y_test)

print("Accuracy:", accuracy)
```

```python
# Confusion matrix

TP = 0
TN = 0
FP = 0
FN = 0

for i in range(len(y_test)):

    if y_pred[i] == "spam" and y_test.iloc[i] == "spam":
        TP += 1

    elif y_pred[i] == "ham" and y_test.iloc[i] == "ham":
        TN += 1

    elif y_pred[i] == "spam" and y_test.iloc[i] == "ham":
        FP += 1

    else:
        FN += 1


print("Confusion Matrix")
print([[TP, FP],
       [FN, TN]])
```

```python
# Evaluate the model

precision = TP / (TP + FP)
recall = TP / (TP + FN)
accuracy = (TP + TN) / (TP + TN + FP + FN)

print("Precision:", precision)
print("Recall:", recall)
print("Accuracy:", accuracy)
```
